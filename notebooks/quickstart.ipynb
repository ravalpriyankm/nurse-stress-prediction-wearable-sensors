# Nurse Stress Prediction: reproducible quickstart

This notebook demonstrates the **software pipeline with entirely synthetic data**. It does not reproduce the original study or measure real nursing stress prediction accuracy. For real data, follow the repository README and download the original dataset from Dryad.

**Academic origin:** University of Essex, CE888: Data Science and Decision Making (2022–2023 academic year). The maintained pipeline and documentation were refreshed in 2026.

```python
from pathlib import Path
from nurse_stress.demo import generate_demo_dataset
from nurse_stress.modelling import train_and_evaluate
```

```python
data_path = Path("data/demo_labeled.csv")
rows = generate_demo_dataset(data_path)
print(f"Created {rows:,} synthetic observations at {data_path}")
```

```python
output_path = Path("results/demo_metrics.json")
report = train_and_evaluate(data_path, output_path, dataset_type="synthetic_demo", n_estimators=40)
print(report["warning"])
print("Training participants:", report["split"]["train_participants"])
print("Test participants:", report["split"]["test_participants"])
```

```python
import pandas as pd
pd.DataFrame({
    name: {"macro_f1": result["macro_f1"], "balanced_accuracy": result["balanced_accuracy"]}
    for name, result in report["models"].items()
}).T
```

**Interpretation:** Synthetic data were created with an artificial relationship between EDA, HR and labels. High results here only indicate that the pipeline can learn a synthetic pattern. Real conclusions require the full source dataset, careful clock checks and participant-disjoint evaluation.
